# 🌾 Kaggriculture — Game Viewer
Run the game and visualize it inline with the built-in renderer.

```python
from kaggle_environments import make

env = make("kaggriculture", configuration={"seed": 42})
env.run(["agent.py", "agent.py"])

# Final scores
obs = env.state[0]['observation']
print(f"Player 0: ${obs['farms'][0]['money']:.0f}")
print(f"Player 1: ${obs['farms'][1]['money']:.0f}")
```

```python
env.render(mode="ipython", width=800, height=600)
```
