# 🌾 Kaggriculture — Game Viewer
Run the game and visualize it inline with the built-in Kaggle renderer.

This produces the same pixel-art farm preview seen on the Kaggriculture competition website.

```python
# Step 1: Run the game
from kaggle_environments import make

env = make("kaggriculture", configuration={"seed": 42})
env.run(["main.py", "main.py"])

# Final scores
obs = env.state[0]['observation']
print(f"Player 0: ${obs['farms'][0]['money']:.0f}")
print(f"Player 1: ${obs['farms'][1]['money']:.0f}")
```

```python
# Step 2: Render the game replay (pixel-art farm view)
env.render(mode="ipython", width=1000, height=700)
```
